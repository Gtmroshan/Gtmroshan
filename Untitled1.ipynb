```python
# ============================================================
# 0. SETUP & EXTRACTION FROM GOOGLE DRIVE
# ============================================================
import os
import random
import json
import numpy as np

import cv2
import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader, TensorDataset
from torchvision import models

from sklearn.feature_selection import mutual_info_classif
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

from skimage.feature import hog
from skimage.filters import gabor

# 1. Mount Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

# 2. Extract ZIP Dataset from Drive to Colab
zip_path = "/content/drive/MyDrive/weed_dataset.zip" # Agar Drive par kisi subfolder me ho toh path yahan change karein
extract_dir = "/content/weed_dataset"

if os.path.exists(zip_path):
    print(f"Extracting dataset from {zip_path}...")
    !unzip -q "{zip_path}" -d "{extract_dir}"
    print("Extraction completed!")
else:
    print(f"Warning: {zip_path} not found. Checking if dataset is already extracted locally...")

# Auto-detect DATA_ROOT (handles nested folders if zip creates subfolders)
DATA_ROOT = extract_dir
if os.path.exists(DATA_ROOT) and not os.path.exists(os.path.join(DATA_ROOT, "train")):
    subfolders = [f for f in os.listdir(DATA_ROOT) if os.path.isdir(os.path.join(DATA_ROOT, f))]
    if subfolders:
        DATA_ROOT = os.path.join(DATA_ROOT, subfolders[0])

print(f"Using DATA_ROOT path: {DATA_ROOT}")


# ============================================================
# 1. CONFIGURATION
# ============================================================

IMG_SIZE = 224

BATCH_SIZE = 32
EPOCHS = 100

LEARNING_RATE = 0.001
WEIGHT_DECAY = 0.005

DROPOUT = 0.4

SEED = 42

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("Using device:", DEVICE)

# Reproducibility
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


# ============================================================
# 2. WN-APO PREPROCESSING
# ============================================================

def calculate_image_quality(image):
    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY).astype(np.float32) / 255.0
    brightness = float(gray.mean())
    noise_variance = float(cv2.Laplacian(gray, cv2.CV_32F).var())

    histogram, _ = np.histogram(gray, bins=256, range=(0, 1), density=True)
    histogram = histogram + 1e-12
    entropy = float(-(histogram * np.log2(histogram)).sum() / 256)

    return brightness, noise_variance, entropy


def wn_apo(image):
    brightness, noise, entropy = calculate_image_quality(image)

    # Adaptive bilateral filtering
    if noise > 0.015:
        sigma_color, sigma_space = 90, 90
    elif noise > 0.006:
        sigma_color, sigma_space = 65, 65
    else:
        sigma_color, sigma_space = 40, 40

    image = cv2.bilateralFilter(image, 5, sigma_color, sigma_space)

    # Adaptive brightness / contrast
    if entropy < 0.55 or brightness < 0.25 or brightness > 0.75:
        lab = cv2.cvtColor(image, cv2.COLOR_RGB2LAB)
        l, a, b = cv2.split(lab)
        clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
        l = clahe.apply(l)
        image = cv2.cvtColor(cv2.merge([l, a, b]), cv2.COLOR_LAB2RGB)

    # Min-Max normalization
    image = image.astype(np.float32)
    minimum, maximum = image.min(), image.max()
    image = (image - minimum) / (maximum - minimum + 1e-8)

    return image


# ============================================================
# 3. DATA AUGMENTATION
# ============================================================

def augment_image(image, mask):
    if random.random() < 0.5:
        image = np.fliplr(image).copy()
        mask = np.fliplr(mask).copy()

    if random.random() < 0.5:
        image = np.flipud(image).copy()
        mask = np.flipud(mask).copy()

    if random.random() < 0.5:
        k = random.choice([1, 2, 3])
        image = np.rot90(image, k).copy()
        mask = np.rot90(mask, k).copy()

    if random.random() < 0.35:
        factor = random.uniform(0.8, 1.2)
        image = np.clip(image * factor, 0, 1)

    return image, mask


# ============================================================
# 4. COCO DATASET LOADER
# ============================================================

TRAIN_DIR = os.path.join(DATA_ROOT, "train")
TEST_DIR = os.path.join(DATA_ROOT, "test")
TRAIN_ANN = os.path.join(TRAIN_DIR, "_annotations.coco.json")
TEST_ANN = os.path.join(TEST_DIR, "_annotations.coco.json")


def load_coco_records(split_dir, annotation_file):
    if not os.path.isdir(split_dir):
        raise FileNotFoundError(f"Missing dataset folder: {split_dir}")
    if not os.path.isfile(annotation_file):
        raise FileNotFoundError(f"Missing COCO annotation file: {annotation_file}")

    with open(annotation_file, "r", encoding="utf-8") as f:
        coco = json.load(f)

    image_info = {item["id"]: item for item in coco["images"]}
    annotations_by_image = {}

    for ann in coco.get("annotations", []):
        annotations_by_image.setdefault(ann["image_id"], []).append(ann)

    records = []
    for image_id, info in image_info.items():
        image_path = os.path.join(split_dir, info["file_name"])
        if not os.path.isfile(image_path):
            continue

        anns = annotations_by_image.get(image_id, [])
        boxes = []
        for ann in anns:
            bbox = ann.get("bbox", None)
            if bbox is None or len(bbox) != 4:
                continue
            x, y, w, h = bbox
            if w > 0 and h > 0:
                boxes.append([float(x), float(y), float(w), float(h)])

        records.append({
            "image_path": image_path,
            "width": int(info["width"]),
            "height": int(info["height"]),
            "boxes": boxes,
        })

    return records


class WeedCocoDataset(Dataset):
    def __init__(self, records, training=False):
        self.records = records
        self.training = training

    def __len__(self):
        return len(self.records)

    def __getitem__(self, index):
        record = self.records[index]

        image = cv2.imread(record["image_path"])
        if image is None:
            raise RuntimeError(f"Cannot read: {record['image_path']}")

        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        original_h, original_w = image.shape[:2]

        mask = np.zeros((original_h, original_w), dtype=np.uint8)

        for x, y, w, h in record["boxes"]:
            x1, y1 = max(0, int(round(x))), max(0, int(round(y)))
            x2, y2 = min(original_w, int(round(x + w))), min(original_h, int(round(y + h)))
            if x2 > x1 and y2 > y1:
                mask[y1:y2, x1:x2] = 255

        image = cv2.resize(image, (IMG_SIZE, IMG_SIZE), interpolation=cv2.INTER_AREA)
        mask = cv2.resize(mask, (IMG_SIZE, IMG_SIZE), interpolation=cv2.INTER_NEAREST)

        image = wn_apo(image)
        mask = (mask > 127).astype(np.float32)

        if self.training:
            image, mask = augment_image(image, mask)

        image = torch.tensor(image.transpose(2, 0, 1), dtype=torch.float32)
        mask = torch.tensor(mask[None, ...], dtype=torch.float32)
        label = torch.tensor(float(mask.max() > 0), dtype=torch.float32)

        return image, mask, label


# ============================================================
# 5. MODEL ARCHITECTURE COMPONENTS
# ============================================================

class ConvBlock(nn.Module):
    def __init__(self, input_channels, output_channels, dilation=1):
        super().__init__()
        padding = dilation
        self.block = nn.Sequential(
            nn.Conv2d(input_channels, output_channels, kernel_size=3, padding=padding, dilation=dilation, bias=False),
            nn.BatchNorm2d(output_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(output_channels, output_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(output_channels),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        return self.block(x)


class AttentionGate(nn.Module):
    def __init__(self, x_channels, g_channels, intermediate_channels):
        super().__init__()
        self.theta = nn.Conv2d(x_channels, intermediate_channels, 1)
        self.phi = nn.Conv2d(g_channels, intermediate_channels, 1)
        self.psi = nn.Conv2d(intermediate_channels, 1, 1)

    def forward(self, x, g):
        g = F.interpolate(g, size=x.shape[-2:], mode="bilinear", align_corners=False)
        attention = self.theta(x) + self.phi(g)
        attention = F.relu(attention)
        attention = torch.sigmoid(self.psi(attention))
        return x * attention


class ASPP(nn.Module):
    def __init__(self, input_channels, output_channels):
        super().__init__()
        dilation_rates = [1, 6, 12, 18]
        self.branches = nn.ModuleList()

        for rate in dilation_rates:
            self.branches.append(
                nn.Sequential(
                    nn.Conv2d(input_channels, output_channels, kernel_size=3, padding=rate, dilation=rate, bias=False),
                    nn.BatchNorm2d(output_channels),
                    nn.ReLU(inplace=True)
                )
            )

        self.project = nn.Sequential(
            nn.Conv2d(output_channels * 4, output_channels, 1),
            nn.BatchNorm2d(output_channels),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        features = [branch(x) for branch in self.branches]
        output = torch.cat(features, dim=1)
        return self.project(output)


class AA_HDN(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder1 = ConvBlock(3, 32)
        self.encoder2 = ConvBlock(32, 64)
        self.encoder3 = ConvBlock(64, 128)
        self.encoder4 = ConvBlock(128, 256)
        self.pool = nn.MaxPool2d(2)

        self.bottleneck = nn.Sequential(
            ConvBlock(256, 512, dilation=2),
            ASPP(512, 512)
        )

        self.attention4 = AttentionGate(256, 512, 128)
        self.attention3 = AttentionGate(128, 256, 64)
        self.attention2 = AttentionGate(64, 128, 32)
        self.attention1 = AttentionGate(32, 64, 16)

        self.decoder4 = ConvBlock(512 + 256, 256)
        self.decoder3 = ConvBlock(256 + 128, 128)
        self.decoder2 = ConvBlock(128 + 64, 64)
        self.decoder1 = ConvBlock(64 + 32, 32)

        self.output = nn.Conv2d(32, 1, kernel_size=1)

    def upsample(self, x, reference):
        return F.interpolate(x, size=reference.shape[-2:], mode="bilinear", align_corners=False)

    def forward(self, x):
        e1 = self.encoder1(x)
        e2 = self.encoder2(self.pool(e1))
        e3 = self.encoder3(self.pool(e2))
        e4 = self.encoder4(self.pool(e3))

        b = self.bottleneck(self.pool(e4))

        a4 = self.attention4(e4, b)
        d4 = self.decoder4(torch.cat([self.upsample(b, e4), a4], dim=1))

        a3 = self.attention3(e3, d4)
        d3 = self.decoder3(torch.cat([self.upsample(d4, e3), a3], dim=1))

        a2 = self.attention2(e2, d3)
        d2 = self.decoder2(torch.cat([self.upsample(d3, e2), a2], dim=1))

        a1 = self.attention1(e1, d2)
        d1 = self.decoder1(torch.cat([self.upsample(d2, e1), a1], dim=1))

        return self.output(d1)


# ============================================================
# 6. LOSSES & METRICS
# ============================================================

def dice_loss(prediction, target):
    smooth = 1e-7
    intersection = (prediction * target).sum(dim=(1, 2, 3))
    denominator = prediction.sum(dim=(1, 2, 3)) + target.sum(dim=(1, 2, 3))
    dice = (2 * intersection + smooth) / (denominator + smooth)
    return (1 - dice).mean()


def segmentation_loss(logits, target):
    prediction = torch.sigmoid(logits)
    bce = F.binary_cross_entropy_with_logits(logits, target)
    dice = dice_loss(prediction, target)
    return bce + dice


def calculate_dice(prediction, target):
    prediction = (prediction > 0.5).float()
    intersection = (prediction * target).sum(dim=(1, 2, 3))
    denominator = prediction.sum(dim=(1, 2, 3)) + target.sum(dim=(1, 2, 3))
    score = (2 * intersection + 1e-7) / (denominator + 1e-7)
    return score.mean().item()


def calculate_iou(prediction, target):
    prediction = (prediction > 0.5).float()
    intersection = (prediction * target).sum(dim=(1, 2, 3))
    union = prediction.sum(dim=(1, 2, 3)) + target.sum(dim=(1, 2, 3)) - intersection
    score = (intersection + 1e-7) / (union + 1e-7)
    return score.mean().item()


# ============================================================
# 7. TRAIN AA-HDN
# ============================================================

def train_aa_hdn(model, train_loader, validation_loader):
    model = model.to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
    scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=20, gamma=0.1)

    best_validation_loss = float("inf")

    for epoch in range(EPOCHS):
        model.train()
        training_loss = 0

        for images, masks, _ in train_loader:
            images, masks = images.to(DEVICE), masks.to(DEVICE)
            optimizer.zero_grad()
            outputs = model(images)
            loss = segmentation_loss(outputs, masks)
            loss.backward()
            optimizer.step()
            training_loss += loss.item()

        model.eval()
        validation_loss = 0
        dice_scores, iou_scores = [], []

        with torch.no_grad():
            for images, masks, _ in validation_loader:
                images, masks = images.to(DEVICE), masks.to(DEVICE)
                outputs = model(images)
                loss = segmentation_loss(outputs, masks)
                validation_loss += loss.item()

                probabilities = torch.sigmoid(outputs)
                dice_scores.append(calculate_dice(probabilities, masks))
                iou_scores.append(calculate_iou(probabilities, masks))

        scheduler.step()
        training_loss /= len(train_loader)
        validation_loss /= len(validation_loader)

        print(
            f"Epoch {epoch+1}/{EPOCHS} | "
            f"Train Loss: {training_loss:.4f} | "
            f"Val Loss: {validation_loss:.4f} | "
            f"Dice: {np.mean(dice_scores):.4f} | "
            f"IoU: {np.mean(iou_scores):.4f}"
        )

        if validation_loss < best_validation_loss:
            best_validation_loss = validation_loss
            torch.save(model.state_dict(), "AA_HDN_best.pth")

    return model


# ============================================================
# 8. FEATURE EXTRACTION & SELECTION (MS-RSFE + PWFS)
# ============================================================

class ResNet50FeatureExtractor(nn.Module):
    def __init__(self):
        super().__init__()
        base_model = models.resnet50(weights=None)
        self.stem = nn.Sequential(base_model.conv1, base_model.bn1, base_model.relu, base_model.maxpool)
        self.layer1, self.layer2 = base_model.layer1, base_model.layer2
        self.layer3, self.layer4 = base_model.layer3, base_model.layer4

    def forward(self, x):
        x = self.stem(x)
        c2 = self.layer1(x)
        c3 = self.layer2(c2)
        c4 = self.layer3(c3)
        c5 = self.layer4(c4)

        p2 = F.adaptive_avg_pool2d(c2, 1).flatten(1)
        p3 = F.adaptive_avg_pool2d(c3, 1).flatten(1)
        p4 = F.adaptive_avg_pool2d(c4, 1).flatten(1)
        p5 = F.adaptive_avg_pool2d(c5, 1).flatten(1)

        return torch.cat([p2, p3, p4, p5], dim=1)


def extract_handcrafted_features(image):
    gray = cv2.cvtColor((image * 255).astype(np.uint8), cv2.COLOR_RGB2GRAY).astype(np.float32) / 255.0

    hog_features = hog(
        gray, orientations=9, pixels_per_cell=(16, 16),
        cells_per_block=(2, 2), block_norm="L2-Hys"
    )

    gabor_features = []
    angles = [0, np.pi / 4, np.pi / 2, 3 * np.pi / 4]
    for angle in angles:
        real, imaginary = gabor(gray, frequency=0.2, theta=angle)
        gabor_features.extend([real.mean(), real.std(), imaginary.mean(), imaginary.std()])

    return np.concatenate([hog_features, np.array(gabor_features, dtype=np.float32)])


def extract_ms_rsfe_features(dataset, segmentation_model):
    segmentation_model.eval()
    resnet = ResNet50FeatureExtractor().to(DEVICE)
    resnet.eval()

    all_features, all_labels = [], []
    loader = DataLoader(dataset, batch_size=1, shuffle=False)

    with torch.no_grad():
        for images, _, labels in loader:
            images = images.to(DEVICE)
            segmentation = torch.sigmoid(segmentation_model(images))[0, 0].cpu().numpy()
            image = images[0].cpu().numpy().transpose(1, 2, 0)

            weed_mask = (segmentation > 0.5)
            masked_image = image * weed_mask[..., None]

            handcrafted = extract_handcrafted_features(masked_image)
            deep_features = resnet(images).cpu().numpy()[0]

            combined_features = np.concatenate([handcrafted, deep_features])
            all_features.append(combined_features)
            all_labels.append(int(labels.item()))

    return np.array(all_features, dtype=np.float32), np.array(all_labels)


def PWFS(X, y, maximum_features=512, correlation_threshold=0.90):
    print("Original features:", X.shape[1])
    X = np.nan_to_num(X)
    X = (X - X.mean(axis=0)) / (X.std(axis=0) + 1e-8)

    mutual_information = mutual_info_classif(X, y, random_state=SEED)
    feature_order = np.argsort(mutual_information)[::-1]

    selected_features = []
    for feature in feature_order:
        if mutual_information[feature] <= 0:
            continue

        if len(selected_features) == 0:
            selected_features.append(feature)
        else:
            correlations = np.corrcoef(X[:, feature], X[:, selected_features].T)[0, 1:]
            if np.all(np.abs(correlations) < correlation_threshold):
                selected_features.append(feature)

        if len(selected_features) >= maximum_features:
            break

    selected_features = np.array(selected_features)
    print("Selected features:", len(selected_features))
    return X[:, selected_features], selected_features


# ============================================================
# 9. CLASSIFIER (WEEDATTNX-NET)
# ============================================================

class TemporalAttention(nn.Module):
    def __init__(self, hidden_size):
        super().__init__()
        self.score = nn.Linear(hidden_size, 1)

    def forward(self, x):
        attention = torch.softmax(self.score(x).squeeze(-1), dim=1)
        weighted = x * attention.unsqueeze(-1)
        return weighted.sum(dim=1)


class WeedAttnXNet(nn.Module):
    def __init__(self, number_of_features):
        super().__init__()
        self.feature_layer = nn.Linear(number_of_features, 256)
        self.rnn = nn.RNN(input_size=256, hidden_size=128, batch_first=True)
        self.bigru = nn.GRU(input_size=128, hidden_size=128, batch_first=True, bidirectional=True)
        self.attention = TemporalAttention(256)
        self.dropout = nn.Dropout(DROPOUT)
        self.classifier = nn.Linear(256, 2)

    def forward(self, x):
        x = self.feature_layer(x).unsqueeze(1).repeat(1, 8, 1)
        x, _ = self.rnn(x)
        x, _ = self.bigru(x)
        x = self.attention(x)
        x = self.dropout(x)
        return self.classifier(x)


def train_classifier(X_train, y_train, X_validation, y_validation):
    X_train = torch.tensor(X_train, dtype=torch.float32)
    y_train = torch.tensor(y_train, dtype=torch.long)
    X_validation = torch.tensor(X_validation, dtype=torch.float32)
    y_validation = torch.tensor(y_validation, dtype=torch.long)

    train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
    validation_loader = DataLoader(TensorDataset(X_validation, y_validation), batch_size=BATCH_SIZE)

    model = WeedAttnXNet(X_train.shape[1]).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
    scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=20, gamma=0.1)
    criterion = nn.CrossEntropyLoss()

    best_loss = float("inf")

    for epoch in range(EPOCHS):
        model.train()
        training_loss = 0
        for X, y in train_loader:
            X, y = X.to(DEVICE), y.to(DEVICE)
            optimizer.zero_grad()
            loss = criterion(model(X), y)
            loss.backward()
            optimizer.step()
            training_loss += loss.item()

        model.eval()
        validation_loss = 0
        with torch.no_grad():
            for X, y in validation_loader:
                X, y = X.to(DEVICE), y.to(DEVICE)
                validation_loss += criterion(model(X), y).item()

        scheduler.step()
        training_loss /= len(train_loader)
        validation_loss /= len(validation_loader)

        print(
            f"Classifier Epoch {epoch+1}/{EPOCHS} | "
            f"Train: {training_loss:.4f} | Validation: {validation_loss:.4f}"
        )

        if validation_loss < best_loss:
            best_loss = validation_loss
            torch.save(model.state_dict(), "WeedAttnXNet_best.pth")

    return model


def evaluate_model(model, X_test, y_test):
    model.eval()
    X_test_tensor = torch.tensor(X_test, dtype=torch.float32).to(DEVICE)

    with torch.no_grad():
        outputs = model(X_test_tensor)
        probabilities = torch.softmax(outputs, dim=1)[:, 1].cpu().numpy()
        predictions = torch.argmax(outputs, dim=1).cpu().numpy()

    print("\n================================")
    print("FINAL RESULTS")
    print("================================")
    print("\nClassification Report:")
    print(classification_report(y_test, predictions, target_names=["Non-Weed", "Weed"], digits=4))

    print("Accuracy :", accuracy_score(y_test, predictions))
    print("Precision:", precision_score(y_test, predictions, zero_division=0))
    print("Recall   :", recall_score(y_test, predictions, zero_division=0))
    print("F1 Score :", f1_score(y_test, predictions, zero_division=0))

    if len(np.unique(y_test)) == 2:
        print("ROC-AUC  :", roc_auc_score(y_test, probabilities))

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, predictions))


# ============================================================
# 10. MAIN RUNNER
# ============================================================

def main():
    print("\n======================================")
    print("AgroWeedX-Ensemble Pipeline Starting...")
    print("======================================\n")

    train_records = load_coco_records(TRAIN_DIR, TRAIN_ANN)
    test_records = load_coco_records(TEST_DIR, TEST_ANN)

    print("COCO training images:", len(train_records))
    print("COCO test images:", len(test_records))

    random_generator = np.random.default_rng(SEED)
    indices = random_generator.permutation(len(train_records))
    val_size = max(1, int(0.20 * len(indices)))

    validation_records = [train_records[i] for i in indices[:val_size]]
    training_records = [train_records[i] for i in indices[val_size:]]

    training_dataset = WeedCocoDataset(training_records, training=True)
    validation_dataset = WeedCocoDataset(validation_records, training=False)
    test_dataset = WeedCocoDataset(test_records, training=False)

    training_loader = DataLoader(training_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
    validation_loader = DataLoader(validation_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    # 1. Train AA-HDN
    print("\nTraining AA-HDN...")
    aa_hdn = AA_HDN()
    aa_hdn = train_aa_hdn(aa_hdn, training_loader, validation_loader)
    aa_hdn.load_state_dict(torch.load("AA_HDN_best.pth", map_location=DEVICE))
    aa_hdn.eval()

    # 2. MS-RSFE Feature Extraction
    print("\nExtracting MS-RSFE features...")
    X_train, y_train = extract_ms_rsfe_features(training_dataset, aa_hdn)
    X_val, y_val = extract_ms_rsfe_features(validation_dataset, aa_hdn)
    X_test, y_test = extract_ms_rsfe_features(test_dataset, aa_hdn)

    # 3. PWFS Selection
    print("\nRunning PWFS Feature Selection...")
    X_train_selected, selected_indices = PWFS(X_train, y_train, maximum_features=512)
    X_val_selected = X_val[:, selected_indices]
    X_test_selected = X_test[:, selected_indices]

    # 4. Train Classifier
    print("\nTraining WeedAttnX-Net Classifier...")
    weed_attnx = train_classifier(X_train_selected, y_train, X_val_selected, y_val)
    weed_attnx.load_state_dict(torch.load("WeedAttnXNet_best.pth", map_location=DEVICE))

    # 5. Evaluate
    evaluate_model(weed_attnx, X_test_selected, y_test)

    # Save Checkpoints
    torch.save(aa_hdn.state_dict(), "AA_HDN_final.pth")
    torch.save(weed_attnx.state_dict(), "WeedAttnXNet_final.pth")
    print("\nPipeline finished! Models saved successfully.")

if __name__ == "__main__":
    main()
```
